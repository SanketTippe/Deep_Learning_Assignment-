## 2. Objective
The main objective is to predict the output variable based on two input features x1 and x2 using a basic neural network model with the equation:
y=3x1+4x2


```python
# importing libraries 
import numpy as np
import matplotlib.pyplot as plt
```

### 1) Activation Functions

Activation functions introduce non-linearity into the neural network, enabling it to learn complex patterns.

### (a) Sigmoid Function

Maps input values between 0 and 1.
### (b) Tanh Function

Maps input values between -1 and 1.

### (c) ReLU Function

Returns input if positive; otherwise returns 0.

```python
def sigmoid(x):
    sig=1/(1+np.exp(-x))
    return sig
def tanh(x):
    tan=np.tanh(x)
    return tan
def relu(x):
    rel=np.maximum(0,x)
    return rel
    
```

### 2) Sample Data: 
We set our input features ( x_1 ) and ( x_2 ) to 3 and 2, respectively, and 
the actual output ( y_{actual} ) to 17.

```python
x1 = 3
x2 = 2
y_actual = 17
```

### 3) Weight Initialization:
We initialize the weights ( w_1 ) and 
( w_2 ) randomly for our model. 
The random seed ensures that we can reproduce the results.

```python
# Initialize weights randomly
np.random.seed(42)  # For reproducibility
w1 = np.random.rand()
w2 = np.random.rand()
```

### 4) Training Process:

We loop through a specified number of epochs (20 in this case).
In each epoch, we perform forward propagation to calculate the predicted output ( y_{pred} ) using the current weights.
We compute the squared error between the actual and predicted values.
### Backpropagation 
is performed to calculate the gradient of the error with respect to the weights, and we update the weights using gradient descent.

```python
# Learning rate
learning_rate = 0.01  # Set the learning rate
epochs = 20           # Number of training iterations
errors = []           # List to store error values for visualization

# Training Process
for epoch in range(epochs):
    # Forward Propagation
    y_pred = x1 * w1 + x2 * w2
```

### Forward Propagation: 
This is the process where input data is passed through the network to generate predictions.
Error Calculation: After predictions are made, we need to measure how far off these predictions are from the actual values. This is typically done using a loss function.
### Backpropagation: 
This is the method used to update the weights of the network based on the error calculated. It involves computing the gradient of the loss function with respect to each weight.

```python
# Training Process
for epoch in range(epochs):
    # Forward Propagation
    y_pred = x1 * w1 + x2 * w2
  # Error Calculation
    error = (y_actual - y_pred) ** 2
    errors.append(error)
    
    # Backpropagation
    d_error = -2 * (y_actual - y_pred)  # Derivative of the error
    w1 -= learning_rate * d_error * x1  # Update weight w1
    w2 -= learning_rate * d_error * x2  # Update weight w2
```

```python
# Visualization
plt.plot(errors)  # Plot the errors to visualize the training process
plt.title('Error Reduction Over Epochs')  # Title of the plot
plt.xlabel('Epochs')  # Label for the x-axis
plt.ylabel('Error')  # Label for the y-axis
plt.show()  # Display the plot

# Final weights and prediction
print(f"Final weights: w1 = {w1}, w2 = {w2}")
print(f"Predicted value: {y_pred}, Actual value: {y_actual}")
```
